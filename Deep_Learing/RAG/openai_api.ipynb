```python
import os
from google import genai

client = genai.Client(
    api_key=os.getenv("GOOGLE_KEY")
)

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents="Write a short bedtime story about a unicorn."
)

print(response.text)
```

```python
from google import genai
from google.genai import types

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents="a + 2 = 7. Find the value of a.",
    config=types.GenerateContentConfig(
        system_instruction="You are a math teacher."
    )
)
```

```python
# with conversation

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=[
        types.Content(
            role="user",
            parts=[types.Part(text="a + b = 7. Find the value of a.")]
        )
    ],
    config=types.GenerateContentConfig(
        system_instruction="You are a math teacher."
    )
)
response
```
