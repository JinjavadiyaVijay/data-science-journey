```python
from dotenv import load_dotenv
import os
load_dotenv()

OPENAI_API_KEY = os.getenv("OPENAI_API_KEY")
```

```python
from openai import OpenAI

client = OpenAI(
    api_key = OPENAI_API_KEY
)
```

```python
class Aiassistant:
    def __init__(self):
        print("hello, i'm your personal Assistant, ask me anything.")

    def response(self):
        question = input("ask me anything?")
        
        response = client.responses.create(
            model = "gpt-5.4",
            input = [
                {"role":"user","content":question},
                {"role":"system","content":"act like persnal ai assistant."}
            ],
            temprature = 1,
            max_output_token = 600
            )
        print(response.text.strip())
    
    def email_summarization (self):
        question = input(" Enter your email to summarize")

        response = client.responses.create(

            model = "gpt-5.4",
            input = [
                        {"role":"user","content":question},
                        {"role":"system","content":"Act as professional Email assistan"}
                    ],
                temprature = 0.3,
                max_output_token = 600
        )
        print(response)
```

```python
assistant = Aiassistant()

assistant.response("hello,")
```

```python
from google import genai
from google.genai import types

gemini_key = os.getenv("GOOGLE_KEY")

client = genai.Client(api_key=gemini_key)



class AIAssistant:
    def __init__(self):
        print("Hello, I'm your personal assistant. Ask me anything!")

    def response(self):
        question = input("Ask me anything: ")

        response = client.models.generate_content(
            model="gemini-2.5-flash",
            contents=question,
            config=types.GenerateContentConfig(
                system_instruction="Act like a personal AI assistant.",
                temperature=1.0,
                max_output_tokens=600,
            ),
        )
        print(response.text)
    
    def email_summarization(self):
        question = input("enter your email")

        response = client.models.generate_content(

            model="gemini-2.5-flash",
            contents=question,
            config = types.GenerateContentConfig(
                system_instruction = "Act like real email assistant.",
                temperature = 0.3,
                max_output_tokens = 600
            ),
        )
        print(response.text.strip())

```

```python
ai_assistant = AIAssistant()

ai_assistant.email_summarization()
```
