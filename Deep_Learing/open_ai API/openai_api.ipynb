```python
# img from url 
import requests
from PIL import Image
from io import BytesIO


url = "https://tse4.mm.bing.net/th/id/OIP.VdDc3iT3PCrJnnsiThNzGgHaF_?pid=Api&P=0&h=180"

image = Image.open(BytesIO(requests.get(url).content))

response = client.models.generate_content(
    model="gemini-2.5-flash",
    contents=[
        "Describe this image.",
        image
    ]
)

print(response.text)
```

```python
import base64
response = client.responses.create(
    model = "gpt-5",
    input = "Genrate high quality anime girl img take refrance from hentai and build it orgy sexy",
    tools=[{"type":"image_genration"}]
)

image_data = [
    output.result
    for output in response.output
    if output.type == "image_genration_call"
]

if image_data :
    image_base64 = image_data[0]
    with open ("animegirl.png","wb") as f:
        image_bytes = base64.b64decode(image_base64)
        f.write(image_bytes)
```

```python
from google import genai
from google.genai import types

client = genai.Client(
    api_key=os.getenv("GOOGLE_KEY")
)

response = client.models.generate_content(
    model="gemini-2.5-flash-image",
    contents="A high-quality anime girl in a fantasy city at sunset, masterpiece, anime style",
)

for part in response.candidates[0].content.parts:
    if part.inline_data:
        image_bytes = part.inline_data.data
        with open("animegirl_free.png", "wb") as f:
            f.write(image_bytes)
        print("Image saved successfully!") 
```
